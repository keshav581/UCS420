```python
#Q3 # (i)
import random


def isPrime(no):
    for i in range (2,no):
        if no%i==0:
            return False
    return True




roll_no=1024170166
random.seed(roll_no)
numbers=[random.randint(100,900) for _ in range(100)]
print(numbers)


# (ii)
odd_numbers=[num for num in numbers if num%2!=0]
print("Odd nos: ",odd_numbers)
print("No of odd no: ",len(odd_numbers))


even_numbers=[num for num in numbers if num%2==0]
print("Even nos: ", even_numbers)
print("No of even no: ",len(even_numbers))


# (iii)
prime_numbers=[num for num in numbers if isPrime(num)]
print("Prime nos: ",prime_numbers)
print("No of prime no",len(prime_numbers))

# (iv)
most_freq_element=max(set(numbers),key=numbers.count)
print("Most frequent element: ",most_freq_element)
print("Count: ",numbers.count(most_freq_element))
```

```python
# Q4
roll_no_str = str(roll_no) # Get the roll number as a string
digits = [int(digit) for digit in roll_no_str]

# Create set A
A = {digit * 7 for digit in digits}

# Create set B
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

# (i) Find the union of A and B
union_AB = A.union(B)
print("Union of A and B:", union_AB)

# (ii) Find the intersection of A and B
intersection_AB = A.intersection(B)
print("Intersection of A and B:", intersection_AB)

# (iii) Find A - B and B - A
difference_A_B = A.difference(B)
difference_B_A = B.difference(A)

print("A - B:", difference_A_B)
print("B - A:", difference_B_A)

#The difference() method finds elements in
#one set but not the other,
#while symmetric_difference() finds elements
#unique to either set (the union of A-B and B-A).

# (iv) Find the symmetric difference of A and B
symmetric_difference_AB = A.symmetric_difference(B)
print("Symmetric Difference of A and B:", symmetric_difference_AB)

# (v) Check whether A is a subset of B, and whether B is a superset of A
is_A_subset_B = A.issubset(B)
is_B_superset_A = B.issuperset(A)

print(f"Is A a subset of B? {is_A_subset_B}")
print(f"Is B a superset of A? {is_B_superset_A}")

# (vi) Ask the user to input a value X and remove X from set A using discard()
X = int(input("Enter a value X to remove from set A: "))
A.discard(X)

print(f"Set A after attempting to discard {X}:", A)

#discard() does not raise an error if the element is not found, unlike remove().
```

```python
# Create the dictionary
my_dict = {
    "name": "Keshav Singh",
    "roll_no": "1024170166",
    "branch": "Computer Science",
    "age": 20,
    "city": "Ambala"
}

print("Original Dictionary:")
print(my_dict)

# i. Rename "city" to "location" using pop()
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming 'city' to 'location':")
print(my_dict)

# ii. Add CGPA
my_dict["cgpa"] = 8.5
print("\nAfter adding CGPA:")
print(my_dict)

# iii. Increase age by 1
my_dict["age"] += 1
print("\nAfter increasing age:")
print(my_dict)

# iv. Delete "branch" using pop() and del in two separate copies

dict1 = my_dict.copy()
removed_branch = dict1.pop("branch")
print("\nAfter deleting 'branch' using pop():")
print(dict1)

dict2 = my_dict.copy()
del dict2["branch"]
print("\nAfter deleting 'branch' using del:")
print(dict2)

# pop() removes the key and returns its value, whereas del only removes the key and returns nothing.

# v. Print all key-value pairs
print("\nKey-Value Pairs:")
for key, value in my_dict.items():
    print(f"{key} -> {value}")

# vi. Check if "email" exists
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")

# vii. Create a friend's dictionary and merge

friend_dict = {
    "name": "Rahul",
    "roll_no": "2311111111",
    "branch": "Information Technology",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("\nMerged Dictionary:")
print(merged_dict)

# When both dictionaries have the same key, the value from the second dictionary (friend_dict) overwrites the first.

# viii. Dictionary containing only string values

string_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}

print("\nDictionary with only string values:")
print(string_dict)
```
