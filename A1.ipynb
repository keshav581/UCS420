```python
prime_sum=0
n=int(input("Enter the no n"))

for num in range(2,n+1):
  is_prime=True
  for i in range(2,num):
    if(num%i==0):
      is_prime=False
  if(is_prime):
    prime_sum=prime_sum+num

prime_sum
```
