```python
prime_sum=0
n=int(input("Enter the no n"))

for num in range(2,n+1):
  is_prime=True
  for i in range(2,num):
    if(num%i==0):
      is_prime=False
  if(is_prime):
    prime_sum=prime_sum+num

prime_sum
```

```python
def Addoddnos(n):
  Sum=0
  return sum(i for i in range(1,n+1) if i%2!=0)

print("Addoddnos(5):",Addoddnos(5))
```

```python
 def Addprimeno(n):
  Sum=0
  for i in range(2,n+1):
    isPrime=True
    for j in range(2,i):
      if(i%j==0): isPrime=False
    if(isPrime): Sum+=i
  return Sum

print("Addprimeno(10):",Addprimeno(10))
```
